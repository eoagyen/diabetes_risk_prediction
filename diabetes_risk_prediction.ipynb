# Diabetes Risk Prediction Using Machine Learning Classification Models

### Background

Diabetes is one of the most common chronic diseases worldwide. Early detection can help patients receive treatment sooner and reduce long-term health complications.

The goal of this project is to develop a machine learning classification system that predicts whether a patient is likely to have diabetes based on clinical and demographic measurements.

The project will compare multiple supervised learning classification algorithms to determine which model provides the best predictive performance.

This project utilizes the DiaBD dataset, a healthcare dataset collected in Bangladesh, to develop and evaluate machine learning classification models capable of predicting diabetes status from demographic, clinical, and health-related indicators.



### Project Objective

##### Main Objective

To build a supervised machine learning classification model that predicts whether an individual is diabetic based on health measurements and medical history.

##### Specific Objectives
- Explore and understand the dataset.
- Identify relationships between risk factors and diabetes.
- Compare multiple classification algorithms.
- Evaluate model performance using healthcare-focused metrics.
- Determine the most influential predictors of diabetes.

### About the Dataset
##### Dataset Description

The DiaBD dataset was introduced in 2025 to support diabetes research, risk assessment, predictive analytics, and healthcare planning in Bangladesh. The dataset contains 5,288 patient records and includes demographic information, clinical measurements, anthropometric indicators, and comorbidity-related variables.
