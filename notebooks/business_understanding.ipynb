# Diabetes Risk Prediction Using Machine Learning Classification Models

## Business Understanding

### Business Problem

Diabetes is one of the most common chronic diseases worldwide. Healthcare providers need effective tools to identify individuals at risk of diabetes as early as possible.

Undiagnosed diabetes can lead to:

- Cardiovascular disease
- Kidney failure
- Vision impairment
- Stroke
- Premature death

Early prediction can support preventive healthcare interventions.



### Project Goal

The goal of this project is to develop a machine learning classification system that predicts whether a patient is likely to have diabetes based on clinical and demographic measurements.

The project will compare multiple supervised learning classification algorithms to determine which model provides the best predictive performance.

This project utilizes the DiaBD dataset, a healthcare dataset collected in Bangladesh, to develop and evaluate machine learning classification models capable of predicting diabetes status from demographic, clinical, and health-related indicators.
